```python
print('L2 norm  ||a||_2 :', la.norm(a))
print('L1 norm  ||a||_1 :', la.norm(a, 1))
print('Linf norm        :', la.norm(a, np.inf))
def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))
print('cosine(a, b)     :', round(cosine(a, b), 3))
```

```python
Pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),
    (np.array([1, 2, 3]), np.array([3, 2, 1])),
    (np.array([2, 0, 0]), np.array([0, 5, 0])),
]
for i, (u, v) in enumerate(Pairs, start=1):
    print(f'Pair {i}:')
    print(f'L1 norm of u: {la.norm(u, 1):.2f}')
    print(f'L2 norm of u: {la.norm(u):.2f}')
    print(f'Cosine similarity of u and v: {cosine(u, v):.3f}')

```

```python
A = np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 0.],
              [4., 2.]])
print('A @ B (matrix multiply):\n', A @ B)
print('A.T   (transpose):\n',A.T)
print('inverse(A):\n', la.inv(A))
print('trace(A) = sum of diagonal:', np.trace(A))
```

```python
I = np.eye(2)
print('Identity I:\n', I)
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))
print('A symmetric?', np.allclose(A, A.T))
theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('Q orthogonal?', np.allclose(Q.T @ Q, I))
```

```python
M= np.array([[4., 2.],
              [2., 3.]])
P = np.array([[0., -1.],
              [1.,  0.]])
print('inverse(M):\n', la.inv(M))
print('trace(M) = sum of diagonal:',np.trace(M))
print('M @ inv(M) == I ?',np.allclose(M @ la.inv(M), I))
print('M symmetric?',np.allclose(M, M.T))
print('P orthogonal?',np.allclose(P.T @ P, I))
```

```python
square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)
S = np.array([[1.5, 0.0],
              [0.0, 0.5]])
t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])
scaled  = S @ square
rotated = R @ square
print('Rotated corners:\n', rotated)
```

```python
def close_loop(pts):
    return np.hstack([pts, pts[:2, :1]])
fig, ax = plt.subplots(figsize=(5, 8))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(rotated),  marker='o', label='rotated 45 deg')
ax.plot(*close_loop(scaled), marker='o', label='scaled')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python
C = np.array([[9.,4.],
              [9.,3.]])
L, V = la.eig(C)
print('Eigenvalues (L):', L)
print('Eigenvectors (V):\n', V)
first_eigenvalue = L[0]
first_eigenvector = V[:, 0]
left_side = C @ first_eigenvector
right_side = first_eigenvalue * first_eigenvector
print(f"C @ v1:\n{left_side}")
print(f"λ1 * v1:\n{right_side}")
print(f"Verification (C v1 == λ1 v1): {np.allclose(left_side, right_side)}")
```
