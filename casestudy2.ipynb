```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
data = {
    'Student': ['S1', 'S2', 'S3', 'S4', 'S5', 'S6', 'S7', 'S8', 'S9', 'S10'],
    'Study_Hours':['2','3', '4', '5','6', '7', '8', '4', '6', '9'],
    'Attendance':['65', '70', '75', '80', '85', '90', '92', '78', '88', '95'],
    'Assignment_Score':['55', '60', '65', '72', '78', '85', '88', '70', '80', '92'],
    'Previous_Score':['58', '62', '68', '70', '76', '82', '86', '72', '79', '90'],
    'Final_Score':['60', '64', '70', '75', '82', '88', '91', '74', '84', '94']
}
df = pd.DataFrame(data)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle(
    "Cohort Performance Metrics Analysis", fontsize=16, fontweight="bold"
)

counts, bins, patches = ax1.hist(
    df["Final_Score"], bins=5, edgecolor="black", color="#3498db", alpha=0.85
)
ax1.set_title("Distribution of Final Scores (5 Bins)", fontsize=12, pad=10)
ax1.set_xlabel("Final Score Interval", fontsize=10)
ax1.set_ylabel("Student Count Density", fontsize=10)
ax1.grid(axis="y", linestyle="--", alpha=0.5)
for count, bin_edge in zip(counts, bins):
    if count > 0:
        ax1.text(
            bin_edge + (bins[1] - bins[0]) / 2,
            count + 0.1,
            str(int(count)),
            ha="center",
            fontweight="bold",
        )

colors = []
for score in df["Final_Score"]:
    if score == df["Final_Score"].min():
        colors.append("#8b0000")
    elif score == df["Final_Score"].max():
        colors.append("#006400")
    else:
        colors.append("#bdc3c7")

bars = ax2.bar(df["Student"], df["Final_Score"], color=colors, edgecolor="black")
ax2.set_title("Final Scores Tracked by Student ID", fontsize=12, pad=10)
ax2.set_xlabel("Student ID", fontsize=10)
ax2.set_ylabel("Final Score", fontsize=10)
ax2.set_ylim(0, 110)
ax2.grid(axis="y", linestyle="--", alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax2.text(
        bar.get_x() + bar.get_width() / 2,
        yval + 1.5,
        str(yval),
        ha="center",
        va="bottom",
        fontweight="bold",
    )

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
def F(x):
    """Predicted final score function."""
    return 40 + 5 * x - 0.2 * x**2

def dF(x):
    """Derivative function (rate of change)."""
    return 5 - 0.4 * x

table_data = []
for x in range(1, 16):
    table_data.append({
        "Study Hours (x)": x,
        "Final Score F(x)": round(F(x), 2),
        "Slope F'(x)": round(dF(x), 2)
    })

df_calc = pd.DataFrame(table_data)
print("--- Numerical Verification Table ---")
print(df_calc.to_string(index=False))
print("\n" + "="*40 + "\n")

x_vals = np.linspace(0, 25, 500)
y_F = F(x_vals)
y_dF = dF(x_vals)

plt.figure(figsize=(10, 6))

plt.plot(x_vals, y_F, label="Predicted Final Score F(x)", color="royalblue", linewidth=2.5)
plt.plot(x_vals, y_dF, label="Rate of Change F'(x)", color="crimson", linewidth=2, linestyle="--")


plt.scatter(12.5, F(12.5), color="darkblue", zorder=5, s=80)
plt.annotate(f"Maximum Score\n(12.5 hrs, {F(12.5):.2f})",
             xy=(12.5, F(12.5)),
             xytext=(14, 68),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6))

plt.scatter(12.5, 0, color="darkred", zorder=5, s=80)
plt.annotate("F'(x) = 0\n(Slope is Flat)",
             xy=(12.5, 0),
             xytext=(14, 5),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6))

plt.title("Optimization of Final Score based on Study Hours", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Study Hours (x)", fontsize=12)
plt.ylabel("Value / Score", fontsize=12)
plt.axhline(0, color='black', linewidth=0.8, linestyle=':')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11, loc="upper right")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd

# Load dataset
data = {
    'Study_Hours': [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],
    'Attendance': [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],
    'Assignment_Score': [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],
    'Final_Score': [60, 64, 70, 75, 82, 88, 91, 74, 84, 94]
}
df = pd.DataFrame(data)

X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values
Y = df["Final_Score"].values

# Add a column of 1s to handle the intercept column (b0)
X = np.column_stack((np.ones(len(X)), X))

# Solve via Normal Equation: beta = (X^T * X)^-1 * X^T * Y
beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("Estimated Coefficients:")
print(f"b0 (Intercept)       : {beta[0]:.4f}")
print(f"b1 (Study Hours)     : {beta[1]:.4f}")
print(f"b2 (Attendance)      : {beta[2]:.4f}")
print(f"b3 (Assignment Score): {beta[3]:.4f}")
```
